```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
```

```python
df = pd.read_csv('titanic_toy.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().mean()
```

```python
X = df.drop(columns=['Survived'])
y = df['Survived']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
X_train.shape, X_test.shape
```

```python
X_train.isnull().mean()
```

```python
mean_age = X_train['Age'].mean()
median_age = X_train['Age'].median()
mean_age = X_train['Fare'].mean()
median_age = X_train['Fare'].median()
```

```python
X_train.sample(5)
```

```python
X_train['Age_median'] = X_train['Age'].fillna(median_age)
X_train['Age_mean'] = X_train['Age'].fillna(mean_age)
X_train['Fare_median'] = X_train['Fare'].fillna(median_age)
X_train['Fare_mean'] = X_train['Fare'].fillna(mean_age)
```

```python
X_train.sample(5)
```

```python
print('Original Age varible varince: ', X_train['Age'].var())
print('Age varince after median imputaion: ', X_train['Age_median'].var())
print('Age varince after mean imputaion: ', X_train['Age_mean'].var())
print('Original Age varible varince: ', X_train['Fare'].var())
print('Fare varince after median imputaion: ', X_train['Fare_median'].var())
print('Fare varince after mean imputaion: ', X_train['Fare_mean'].var())
```

```python
fig = plt.figure()
ax = fig.add_subplot(111)

#original varible distribution
X_train['Fare'].plot(kind='kde', ax=ax)
#variable imputed with the median
X_train['Fare_median'].plot(kind='kde', ax=ax, color='red')
X_train['Fare_mean'].plot(kind='kde', ax=ax, color='green')
lines,lables = ax.get_legend_handles_labels()
ax.legend(lines,lables,loc='best')
```

```python
fig = plt.figure()
ax = fig.add_subplot(111)

#original varible distribution
X_train['Age'].plot(kind='kde', ax=ax)
#variable imputed with the median
X_train['Age_median'].plot(kind='kde', ax=ax, color='red')
X_train['Age_mean'].plot(kind='kde', ax=ax, color='green')
lines,lables = ax.get_legend_handles_labels()
ax.legend(lines,lables,loc='best')
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
imputer1 = SimpleImputer(strategy='median')
imputer2 = SimpleImputer(strategy='mean')
```

```python
trf = ColumnTransformer([
    ('imputer1',imputer1,['Age']),
    ('imputer2',imputer2,['Fare'])
],remainder='passthrough')
```

```python
trf.fit(X_train)
```

```python
trf.named_transformers_['imputer1'].statistics_
```

```python
trf.named_transformers_['imputer2'].statistics_
```

```python
X_train = trf.transform(X_train)
X_test = trf.transform(X_test)
```

```python
# arbitrary Value imputation
```

```python
X_train['Age_99'] = X_train['Age'].fillna(99)
X_train['Age_minus1'] = X_train['Age'].fillna(-1)
X_train['Fare_999'] = X_train['Fare'].fillna(999)
X_train['Fare_minus1'] = X_train['Fare'].fillna(-1)
```

```python
print('Original Age varible varince: ', X_train['Age'].var())
print('Age varince after median imputaion: ', X_train['Age_99'].var())
print('Age varince after mean imputaion: ', X_train['Age_minus1'].var())
print('Original Age varible varince: ', X_train['Fare'].var())
print('Fare varince after median imputaion: ', X_train['Fare_999'].var())
print('Fare varince after mean imputaion: ', X_train['Fare_minus1'].var())
```

```python
fig = plt.figure()
ax = fig.add_subplot(111)

#original varible distribution
X_train['Age'].plot(kind='kde', ax=ax)
#variable imputed with the median
X_train['Age_99'].plot(kind='kde', ax=ax, color='orange')
X_train['Age_minus1'].plot(kind='kde', ax=ax, color='green')
lines,lables = ax.get_legend_handles_labels()
ax.legend(lines,lables,loc='best')
```

```python
fig = plt.figure()
ax = fig.add_subplot(111)

#original varible distribution
X_train['Fare'].plot(kind='kde', ax=ax)
#variable imputed with the median
X_train['Fare_999'].plot(kind='kde', ax=ax, color='red')
X_train['Fare_minus1'].plot(kind='kde', ax=ax, color='green')
lines,lables = ax.get_legend_handles_labels()
ax.legend(lines,lables,loc='best')
```
